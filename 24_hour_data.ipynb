```python
import numpy as np
import pandas as pd 
```

```python
data = pd.read_csv(r"C:\Users\Chand\Downloads\24-hour with cate and placement=1 (1).csv")
```

```python
# add the price paid for the click (if any)
data["price_paid_click"] = data["AUCTION_PRICE"] * np.where(data["CLICK_ID"].notnull(), 1, 0)
data["price_paid_click"] = data["price_paid_click"].fillna(0)
```

```python
import pandas as pd
import numpy as np

data['CLICK_TIME'] = pd.to_datetime(data['CLICK_TIME'], errors='coerce')
data['IMPRESSION_TIME'] = pd.to_datetime(data['IMPRESSION_TIME'], errors='coerce')
# --- 2. Define Custom Aggregation Functions ---
def proportion_non_null(series):
    """Calculates the proportion of non-null (or not-NaN) values in a Series."""
    return series.notna().mean()

def count_non_null(series):
    """Calculates the count of non-null (or not-NaN) values in a Series."""
    return series.notna().sum()

summary_stats = data.groupby('PRODUCT_ID_N').agg(
    # --- Proportions ---
    Win_Proportion=('IS_WINNER', 'mean'),
    Click_Proportion=('CLICK_TIME', proportion_non_null),
    Impression_Proportion=('IMPRESSION_TIME', proportion_non_null),
    
    # --- Counts ---
    Total_Auctions=('AUCTION_ID_N', 'count'), 
    Impression_Count=('IMPRESSION_TIME', count_non_null),
    Click_Count=('CLICK_TIME', count_non_null),
    
    # --- Min/Max Click Time ---
    Min_Click_Time=('CLICK_TIME', 'min'),
    Max_Click_Time=('CLICK_TIME', 'max'),
    
    # --- New: Budget and Price ---
    daily_budget=('price_paid_click', 'sum'),
    mean_catalog_price=('CATALOG_PRICE', 'mean')
).reset_index()



# --- 4. Format the Proportions (for display) ---
summary_stats['Win_Proportion'] = summary_stats['Win_Proportion'].map('{:.2%}'.format)
summary_stats['Click_Proportion'] = summary_stats['Click_Proportion'].map('{:.2%}'.format)
summary_stats['Impression_Proportion'] = summary_stats['Impression_Proportion'].map('{:.2%}'.format)
```
